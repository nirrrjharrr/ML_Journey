## Agglomerative Clustering (Hierarchical Clustering)

`Agglomerative Clustering` is a **hierarchical, unsupervised learning algorithm**
used for clustering data without requiring predefined labels.
It follows a **bottom-up approach**, where each data point starts as its own cluster
and clusters are merged step by step.

It is widely used when the **hierarchical structure of data** is important.

---

### Why Agglomerative Clustering Is Needed

Some clustering problems:
- Do not have a clear number of clusters  
- Require understanding relationships between clusters  
- Have non-spherical cluster shapes  

Agglomerative clustering helps by:
- Revealing cluster hierarchy  
- Not requiring a fixed number of clusters upfront  
- Working well with various distance measures  

---

### Core Idea of Agglomerative Clustering

Agglomerative clustering works by:
- Treating each data point as a separate cluster  
- Iteratively merging the **closest clusters**
- Continuing until all points form a single cluster or a stopping condition is met  

The result is a **hierarchical tree structure**.

---

### How Agglomerative Clustering Works

1. Start with each data point as its own cluster  
2. Compute distances between all clusters  
3. Merge the two closest clusters  
4. Update distances between new cluster and others  
5. Repeat steps 2–4 until desired clusters are formed  

---

### Linkage Criteria

Linkage defines how the distance between clusters is measured.

#### Single Linkage
- Distance between closest points in clusters  
- Can form long chain-like clusters  

#### Complete Linkage
- Distance between farthest points in clusters  
- Produces compact clusters  

#### Average Linkage
- Average distance between all points in clusters  
- Balanced approach  

#### Ward’s Linkage
- Minimizes variance within clusters  
- Most commonly used with Euclidean distance  

---

### Distance Metrics

Common distance measures:
- Euclidean  
- Manhattan  
- Cosine  

Choice of distance greatly affects clustering results.

---

### Dendrogram (Hierarchy Visualization)

- A dendrogram visualizes how clusters merge  
- Height represents distance between merged clusters  
- Cutting the dendrogram at a certain height gives final clusters  

---

### Choosing the Number of Clusters

- Cut the dendrogram at an appropriate height  
- Use domain knowledge  
- Use silhouette score for validation  

---

### Importance of Feature Scaling

- Distance-based algorithm  
- Unscaled features distort distance calculations  
- Standardization is strongly recommended  

---

### Advantages of Agglomerative Clustering

- Does not require specifying number of clusters initially  
- Produces interpretable hierarchical structure  
- Flexible distance and linkage choices  
- Works well with small to medium datasets  

---

### Limitations of Agglomerative Clustering

- Computationally expensive (O(n²))  
- Not suitable for very large datasets  
- Sensitive to noise and outliers  
- No reassignment once clusters are merged  

---

### Agglomerative Clustering vs K-Means

| Aspect | Agglomerative | K-Means |
|------|--------------|--------|
| Requires K upfront | No | Yes |
| Cluster shape | Flexible | Spherical |
| Scalability | Low | High |
| Hierarchy | Yes | No |

---

### When to Use Agglomerative Clustering

Use Agglomerative Clustering when:
- Dataset is small to medium  
- Hierarchical relationships matter  
- Cluster shape is non-spherical  

Avoid when:
- Dataset is very large  
- Speed is critical  

---

### Practical Notes

- Ward linkage + Euclidean distance is a strong default  
- Always scale features before clustering  
- Dendrograms help interpret cluster structure  

---

### Key Takeaways

- Agglomerative clustering is a bottom-up hierarchical method  
- Builds clusters by progressively merging closest groups  
- Produces a dendrogram for interpretability  
- Powerful for exploratory and hierarchical analysis  

Agglomerative clustering is especially useful in exploratory data analysis
and domains where understanding relationships between clusters is critical.


```python
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline
import numpy as np
```

```python
customer_data = pd.read_csv('../../hierarchical-clustering-with-python-and-scikit-learn-shopping-data.csv')
```

```python
customer_data.shape
```

```python
data = customer_data.iloc[:, 3:5].values
data
```

```python
import scipy.cluster.hierarchy as shc

#plt.figure(figsize=(10, 7))
plt.title("Customer Dendograms")
dend = shc.dendrogram(shc.linkage(data, method='ward'))
```

```python
from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(
    n_clusters=5,
    linkage='ward'
)

labels_ = cluster.fit_predict(data)
labels_
```

```python
#plt.figure(figsize=(10, 7))
plt.scatter(data[:,0], data[:,1], c=cluster.labels_, cmap='rainbow')
```
