## Elastic Net Regression

`Elastic Net Regression` is a **regularized linear regression technique**
that combines the strengths of **Ridge Regression** and **Lasso Regression**.
It is designed to handle situations where features are **highly correlated**
and feature selection is still important.

---

### What It Is

Elastic Net:
- Is a linear regression model with regularization
- Combines Ridge and Lasso penalties
- Shrinks coefficients and can remove some features
- Balances stability and feature selection

---

### Why It Is Used

Elastic Net is used because:
- Ridge keeps all features but lacks selection
- Lasso selects features but struggles with correlation
- Elastic Net provides the best of both worlds
- It improves model reliability

---

### Where It Is Used

Elastic Net is commonly used in:
- High-dimensional datasets
- Genomics and bioinformatics
- Financial modeling
- Text and sparse data regression

---

### When It Performs Well

Elastic Net performs well when:
- Features are highly correlated
- Dataset has many features
- Feature selection and stability are both important
- Overfitting needs to be controlled

---

### When It Performs Poorly

Elastic Net performs poorly when:
- Dataset is very small
- Data relationships are highly non-linear
- Hyperparameters are not tuned
- Feature scaling is not applied

---

### Advantages

- Handles correlated features well
- Performs feature selection
- Produces stable and interpretable models
- Flexible regularization

---

### Limitations

- Requires tuning of multiple parameters
- Assumes linear relationships
- Needs feature scaling
- More complex than Ridge or Lasso alone

---

### Key Takeaways

- Elastic Net combines Ridge and Lasso
- Balances feature selection and stability
- Works well with correlated features
- Ideal for high-dimensional regression problems


```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Ridge 
reg = Ridge(alpha=0.1)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Lasso
reg = Lasso(alpha=0.01)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# ElasticNet
reg = ElasticNet(alpha=0.005,l1_ratio=0.9)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```
