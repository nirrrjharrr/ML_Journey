## Mini-Batch Gradient Descent

`Mini-Batch Gradient Descent` is a **hybrid optimization method**
that combines the benefits of **Batch Gradient Descent** and
**Stochastic Gradient Descent (SGD)**.

It updates model parameters using a **small batch of training samples**
instead of the entire dataset or a single data point.

---

### What It Is

Mini-Batch Gradient Descent:
- Divides the dataset into small batches
- Updates model parameters after each batch
- Uses a fixed number of samples per update

It balances speed and stability.

---

### Why It Is Used

Mini-Batch Gradient Descent is used because:
- It is faster than Batch Gradient Descent
- It is more stable than SGD
- It works efficiently with modern hardware (GPUs)
- It is the most commonly used optimization method

---

### Where It Is Used

Mini-Batch Gradient Descent is widely used in:
- Neural networks
- Deep learning models
- Logistic regression
- Large-scale machine learning systems

---

### When It Performs Well

Mini-Batch Gradient Descent performs well when:
- Dataset is large
- Hardware acceleration is available
- Stable and fast convergence is needed
- Training deep models

---

### When It Performs Poorly

Mini-Batch Gradient Descent performs poorly when:
- Batch size is poorly chosen
- Dataset is very small
- Memory is limited
- Improper learning rate is used

---

### Advantages

- Faster than Batch Gradient Descent
- More stable than SGD
- Efficient memory usage
- Well-suited for parallel computation

---

### Limitations

- Requires tuning batch size
- Slightly more complex than SGD
- Still sensitive to learning rate
- Can overfit if not regularized

---

### Key Takeaways

- Uses small batches of data
- Best balance between speed and stability
- Most commonly used gradient descent variant
- Standard choice in deep learning


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
X,y = load_diabetes(return_X_y=True)
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

### Creating Own Mini Batch Gradient Descent Class

```python
import random

class MBGDRegressor:
    
    def __init__(self,batch_size,learning_rate=0.01,epochs=100):
        
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        
    def fit(self,X_train,y_train):
        # init your coefs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            
            for j in range(int(X_train.shape[0]/self.batch_size)):
                
                idx = random.sample(range(X_train.shape[0]),self.batch_size)
                
                y_hat = np.dot(X_train[idx],self.coef_) + self.intercept_
                #print("Shape of y_hat",y_hat.shape)
                intercept_der = -2 * np.mean(y_train[idx] - y_hat)
                self.intercept_ = self.intercept_ - (self.lr * intercept_der)

                coef_der = -2 * np.dot((y_train[idx] - y_hat),X_train[idx])
                self.coef_ = self.coef_ - (self.lr * coef_der)
        
        print(self.intercept_,self.coef_)
    
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
mbr = MBGDRegressor(batch_size=int(X_train.shape[0]/50),learning_rate=0.01,epochs=100)
```

```python
mbr.fit(X_train,y_train)
```

```python
y_pred = mbr.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```
