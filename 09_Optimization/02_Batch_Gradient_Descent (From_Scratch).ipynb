## Batch Gradient Descent

`Batch Gradient Descent` is a **type of Gradient Descent**
where the model parameters are updated using the **entire training dataset**
at each step.

It is the most straightforward and stable form of gradient descent.

---

### What It Is

Batch Gradient Descent:
- Uses all training data to compute each update
- Updates model parameters once per iteration
- Follows a smooth and stable path toward the minimum error

---

### Why It Is Used

Batch Gradient Descent is used because:
- Updates are stable and consistent
- It converges smoothly
- It is easy to understand and implement

---

### Where It Is Used

Batch Gradient Descent is commonly used in:
- Linear regression (small to medium datasets)
- Educational and learning examples
- Convex optimization problems

---

### When It Performs Well

Batch Gradient Descent performs well when:
- Dataset is small to medium
- Memory is sufficient
- Stable convergence is desired
- Loss function is convex

---

### When It Performs Poorly

Batch Gradient Descent performs poorly when:
- Dataset is very large
- Memory is limited
- Fast updates are required
- Real-time learning is needed

---

### Advantages

- Stable and smooth convergence
- Predictable behavior
- Easy to debug and visualize

---

### Limitations

- Slow for large datasets
- High memory usage
- Updates happen infrequently
- Not suitable for online learning

---

### Key Takeaways

- Uses the full dataset for each update
- Stable but slow
- Best for small datasets
- Simple and reliable optimization method


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
```

```python
from sklearn.model_selection import train_test_split
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train.shape
```

### Creating Own Batch Gradient Descent Class

```python
class GDRegressor:
    
    def __init__(self,learning_rate=0.01,epochs=100):
        
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs
        
    def fit(self,X_train,y_train):
        # init your coefs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            # update all the coef and the intercept
            y_hat = np.dot(X_train,self.coef_) + self.intercept_
            #print("Shape of y_hat",y_hat.shape)
            intercept_der = -2 * np.mean(y_train - y_hat)
            self.intercept_ = self.intercept_ - (self.lr * intercept_der)
            
            coef_der = -2 * np.dot((y_train - y_hat),X_train)/X_train.shape[0]
            self.coef_ = self.coef_ - (self.lr * coef_der)
        
        print(self.intercept_,self.coef_)
    
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
gdr = GDRegressor(epochs=1000,learning_rate=0.5)
```

```python
gdr.fit(X_train,y_train)
```

```python
y_pred = gdr.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```
