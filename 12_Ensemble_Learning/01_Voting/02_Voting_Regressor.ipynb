## Voting Regressor

A `Voting Regressor` is an **ensemble learning technique** that combines
predictions from multiple regression models to produce a single final prediction.
Instead of trusting one model, it averages the outputs of several models.

---

### What It Is

A Voting Regressor:
- Trains multiple regression models independently
- Collects their predicted values
- Computes the average prediction as the final output

Each model acts as a “voter” contributing a numeric value.

---

### Why It Is Used

Voting Regressors are used because:
- Averaging predictions reduces individual model errors
- Different models capture different patterns
- Ensemble predictions are more stable and reliable

---

### Where It Is Used

Voting Regressors are commonly used in:
- Regression problems
- Price prediction
- Demand forecasting
- Kaggle competitions
- Baseline ensemble models

---

### When It Performs Well

Voting Regressors perform well when:
- Base models are diverse
- Individual models perform reasonably well
- Dataset has moderate noise
- Relationships are complex

---

### When It Performs Poorly

Voting Regressors perform poorly when:
- Base models are weak
- Models are highly correlated
- Dataset is very small
- One model dominates others

---

### Advantages

- Improves prediction stability
- Reduces variance
- Easy to implement
- Works with different types of models

---

### Limitations

- Increased computation cost
- Less interpretable than single models
- Does not reduce bias significantly
- Requires careful model selection

---

### Key Takeaways

- Voting Regressor averages predictions
- Combines multiple regression models
- Works best with diverse models
- Improves robustness and stability


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()

X = housing.data
y = housing.target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
     
```

```python
estimators = [('lr',lr),('dt',dt),('svr',svr)]
```

```python
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

```python
from sklearn.ensemble import VotingRegressor
     
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```

```python
for i in range(1,4):
  for j in range(1,4):
    for k in range(1,4):
      vr = VotingRegressor(estimators,weights=[i,j,k])
      scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
      print("For i={},j={},k={}".format(i,j,k),np.round(np.mean(scores),2))
```

### Regressor of Same Algo

```python
dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
     
```

```python
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
```

```python
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
     
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```
