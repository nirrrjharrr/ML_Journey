## Voting Regressor (Ensemble Learning)

A `Voting Regressor` is an ensemble learning technique used for **regression problems**.
It combines predictions from multiple regression models and produces a final prediction
by **averaging their outputs**.

The idea is that combining multiple models reduces variance and improves generalization.

---

### Why Use a Voting Regressor

Voting regressors are useful because:
- Different models capture different patterns  
- Individual model errors can cancel out  
- Predictions become more stable and robust  

They work best when base regressors are:
- Diverse (different algorithms)
- Reasonably accurate on their own

---

### How Voting Regressor Works

Steps:
1. Train multiple regression models  
2. Each model predicts a numerical value  
3. Combine predictions using averaging  
4. Output the final prediction  

---

### Types of Voting in Regression

Unlike classification, regression uses **averaging** instead of majority voting.

#### 1. Simple Averaging

The final prediction is the mean of predictions:

$$
\hat{y}_{final} = \frac{1}{n} \sum_{i=1}^{n} \hat{y}_i
$$

All models contribute equally.

---

#### 2. Weighted Averaging

Better-performing models are given higher weights:

$$
\hat{y}_{final} = \sum_{i=1}^{n} w_i \hat{y}_i
$$

Where:
- \( w_i \) = weight of model \( i \)
- \( \sum w_i = 1 \)

---

### Example

Assume three regressors predict house price:

| Model | Prediction |
|------|-----------|
| Model 1 | 200 |
| Model 2 | 220 |
| Model 3 | 210 |

Final prediction (simple averaging):

$$
\frac{200 + 220 + 210}{3} = 210
$$

---

### When to Use Weighted Voting

Weighted voting is useful when:
- Models have different performance levels  
- Some models are more reliable  
- Cross-validation scores are available  

---

### Advantages of Voting Regressor

- Simple and easy to implement  
- Reduces variance  
- Improves stability  
- Less sensitive to overfitting  

---

### Limitations of Voting Regressor

- No learning between models  
- Weak models can hurt performance  
- Increased computation  
- Cannot capture complex interactions  

---

### Voting Regressor vs Individual Models

| Aspect | Individual Model | Voting Regressor |
|------|------------------|------------------|
| Variance | High | Lower |
| Bias | Depends | Similar |
| Stability | Lower | Higher |

---

### Voting Regressor vs Other Ensembles

| Method | Learns Weights | Model Dependency |
|------|----------------|------------------|
| Voting Regressor | No | Independent |
| Random Forest | Yes | Trees only |
| Boosting | Yes | Sequential |
| Stacking | Yes | Meta-learner |

---

### When NOT to Use Voting Regressor

- Base models are very similar  
- Dataset is extremely small  
- One model is clearly dominant  

---

### Practical Notes

- Base regressors do not need the same scale  
- Feature preprocessing should be consistent  
- Works well as a strong baseline  

---

### Key Takeaways

- Voting Regressor combines predictions using averaging  
- Simple yet effective ensemble method  
- Best when models are diverse and individually strong  
- Often used before moving to advanced ensemble techniques  

Voting Regressor is a powerful baseline ensemble for regression tasks and helps
build intuition before learning boosting and stacking methods.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()

X = housing.data
y = housing.target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
     
```

```python
estimators = [('lr',lr),('dt',dt),('svr',svr)]
```

```python
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

```python
from sklearn.ensemble import VotingRegressor
     
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```

```python
for i in range(1,4):
  for j in range(1,4):
    for k in range(1,4):
      vr = VotingRegressor(estimators,weights=[i,j,k])
      scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
      print("For i={},j={},k={}".format(i,j,k),np.round(np.mean(scores),2))
```

### Regressor of Same Algo

```python
dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
     
```

```python
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
```

```python
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
     
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```
